```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
df = pd.read_csv("linear_regression_dataset (2).csv")
print("Dataset Shape:", df.shape)
print(df.head())
```

```python
plt.scatter(df['response_delay'], df['fraud_risk_score'], color='blue', alpha=0.5)
plt.xlabel("Response Delay")
plt.ylabel("Fraud Risk Score")
plt.title("Response Delay vs Fraud Risk Score")
plt.show()
```

```python
X = df[['response_delay']]
y = df['fraud_risk_score']
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

print("Training samples:", len(X_train))
print("Testing samples :", len(X_test))
```

```python
model = LinearRegression()
model.fit(X_train, y_train)

print("\nCoefficient (slope)  :", model.coef_[0])
print("Intercept            :", model.intercept_)

y_pred = model.predict(X_test)

accuracy = model.score(X_test, y_test)
print(f"\nModel Accuracy (R² Score): {accuracy * 100:.2f}%")
```

```python
plt.scatter(X_test, y_test,  color='blue',  label='Actual',    alpha=0.6)
plt.scatter(X_test, y_pred,  color='red',   label='Predicted', alpha=0.6)
plt.plot(X_test, y_pred,     color='green', linewidth=2,       label='Regression Line')
plt.xlabel("Response Delay")
plt.ylabel("Fraud Risk Score")
plt.title(f"Actual vs Predicted  |  Accuracy: {accuracy*100:.2f}%")
plt.legend()
plt.show()

new_value = [[15.0]]
predicted = model.predict(new_value)
print(f"\nPredicted Fraud Risk Score for response_delay={new_value[0][0]}: {predicted[0]:.2f}")
```
