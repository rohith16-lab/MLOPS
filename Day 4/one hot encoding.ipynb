```python
import pandas as pd

# Load dataset
df = pd.read_csv("Students suspicious behaviors detection dataset_V1.csv")

# Encode separately
head_pose_encoded = pd.get_dummies(df['head_pose'], prefix='head_pose')

gaze_direction_encoded = pd.get_dummies(df['gaze_direction'], prefix='gaze')

# Combine with original dataset
df = pd.concat([df, head_pose_encoded, gaze_direction_encoded], axis=1)

# Remove original categorical columns
df.drop(['head_pose', 'gaze_direction'], axis=1, inplace=True)

# Save final dataset
df.to_csv("encoded_dataset.csv", index=False)

print(df.head())
```
